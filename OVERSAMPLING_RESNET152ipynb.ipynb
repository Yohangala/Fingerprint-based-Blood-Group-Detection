```python
# ==============================================================================
# Blood Group Prediction from Fingerprints: Complete Training Notebook
# ==============================================================================
# This script is designed for Google Colab. It uses scikit-learn's resample
# function to oversample the ENTIRE dataset BEFORE splitting.
#
# It implements:
# - ResNet152 with transfer learning
# - Advanced, domain-specific data augmentation (Elastic Distortions)
# - Handling of severe class imbalance (Oversampling with sklearn.utils.resample)
# - A complete training, validation, and evaluation pipeline.
# - An inference function to test on new, random images.
# ==============================================================================

# --- 0. SETUP AND IMPORTS ---
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms
from torchvision.models import ResNet152_Weights

import os
import numpy as np
import time
import copy
import random
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample # ADDED for oversampling
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Albumentations for advanced data augmentation
import albumentations as A
from albumentations.pytorch import ToTensorV2

# Mount Google Drive to access the dataset
try:
    from google.colab import drive
    drive.mount('/content/drive')
    IN_COLAB = True
    print("Google Drive mounted successfully.")
except ImportError:
    IN_COLAB = False
    print("Not running in Google Colab. Assuming local file paths.")


# --- 1. CONFIGURATION ---
class Config:
    # --- IMPORTANT: UPDATE THESE PATHS ---
    # Set the path to the root folder of your dataset in Google Drive
    DATA_DIR = '/content/drive/MyDrive/major/dataset_blood_group'

    # Set the path to save model checkpoints and results
    OUTPUT_DIR = '/content/drive/MyDrive/DATASET'

    # --- Model & Training Parameters ---
    NUM_CLASSES = 8  # A+, A-, B+, B-, AB+, AB-, O+, O-
    BATCH_SIZE = 8
    NUM_EPOCHS = 50
    LEARNING_RATE = 1e-4

    # --- Dataset Parameters ---
    CLASS_NAMES = ['A+', 'A-', 'B+', 'B-', 'AB+', 'AB-', 'O+', 'O-']
    IMAGE_SIZE = 224

    # --- Other Settings ---
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    PATIENCE = 5

# Instantiate the configuration
config = Config()

# Create output directory if it doesn't exist
os.makedirs(config.OUTPUT_DIR, exist_ok=True)
print(f"Using device: {config.DEVICE}")
print(f"Output will be saved to: {config.OUTPUT_DIR}")


# --- 2. DATA AUGMENTATION & TRANSFORMATION ---
def get_train_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.RandomRotate90(p=0.5),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.ElasticTransform(alpha=1, sigma=50, alpha_affine=50, p=0.5),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

def get_val_test_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])


# --- 3. DATASET & DATALOADERS (UPDATED FOR OVERSAMPLING BEFORE SPLIT) ---
class FingerprintDataset(Dataset):
    """Custom PyTorch Dataset for loading fingerprint images."""
    def __init__(self, image_paths, labels, class_to_idx, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.class_to_idx = class_to_idx

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        image = Image.open(image_path).convert("RGB")
        image = np.array(image)
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image=image)["image"]

        return image, label

def prepare_dataloaders():
    """Scans the data directory, oversamples the entire dataset, splits it, and creates DataLoaders."""
    print("Preparing dataloaders...")

    if not os.path.isdir(config.DATA_DIR):
        raise FileNotFoundError(f"Data directory not found at: {config.DATA_DIR}. Please update the path in the Config class.")

    # Create a DataFrame of all images and labels
    filepaths = []
    labels = []
    for class_name in config.CLASS_NAMES:
        class_dir = os.path.join(config.DATA_DIR, class_name)
        if not os.path.isdir(class_dir):
            continue
        for img_name in os.listdir(class_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
                filepaths.append(os.path.join(class_dir, img_name))
                labels.append(class_name)

    data = pd.DataFrame({'Filepath': filepaths, 'Label': labels})

    # --- VISUALIZE ORIGINAL DISTRIBUTION ---
    print("\n--- Original Full Dataset Distribution ---")
    print(data['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Original Full Dataset')
    plt.ylabel('Number of Images')
    plt.show()

    # --- OVERSAMPLING LOGIC (APPLIED TO ENTIRE DATASET) ---
    # WARNING: This approach can lead to data leakage and overly optimistic results.
    # It is generally recommended to split the data first, then oversample only the training set.
    print("\n--- Applying Oversampling to Entire Dataset ---")
    max_count = data['Label'].value_counts().max()
    print(f"The majority class has {max_count} samples. Oversampling minority classes to match this count.")

    class_dfs = [data[data['Label'] == label] for label in config.CLASS_NAMES]

    oversampled_dfs = [
        resample(df, replace=True, n_samples=max_count, random_state=42) if len(df) < max_count else df
        for df in class_dfs
    ]

    data_oversampled = pd.concat(oversampled_dfs)
    data_oversampled = data_oversampled.sample(frac=1, random_state=42).reset_index(drop=True)

    # --- VISUALIZE BALANCED DISTRIBUTION ---
    print("\n--- Full Dataset Distribution After Oversampling ---")
    print(data_oversampled['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data_oversampled, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Full Dataset After Oversampling')
    plt.ylabel('Number of Images')
    plt.show()

    # Split the now-oversampled data into training, validation, and test sets
    train_df, test_df = train_test_split(data_oversampled, test_size=0.30, random_state=42, stratify=data_oversampled['Label'])
    val_df, test_df = train_test_split(test_df, test_size=0.50, random_state=42, stratify=test_df['Label'])

    # Get file paths and labels from the dataframes
    class_to_idx = {name: i for i, name in enumerate(config.CLASS_NAMES)}
    idx_to_class = {i: name for i, name in enumerate(config.CLASS_NAMES)}

    train_paths = train_df['Filepath'].tolist()
    train_labels = [class_to_idx[label] for label in train_df['Label']]

    val_paths = val_df['Filepath'].tolist()
    val_labels = [class_to_idx[label] for label in val_df['Label']]

    test_paths = test_df['Filepath'].tolist()
    test_labels = [class_to_idx[label] for label in test_df['Label']]

    print(f"\nTraining set size: {len(train_paths)}")
    print(f"Validation set size: {len(val_paths)}")
    print(f"Test set size: {len(test_paths)}")

    # Create datasets
    train_dataset = FingerprintDataset(train_paths, train_labels, class_to_idx, transform=get_train_transforms())
    val_dataset = FingerprintDataset(val_paths, val_labels, class_to_idx, transform=get_val_test_transforms())
    test_dataset = FingerprintDataset(test_paths, test_labels, class_to_idx, transform=get_val_test_transforms())

    # Create dataloaders (no sampler needed now)
    train_loader = DataLoader(train_dataset, batch_size=config.BATCH_SIZE, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)

    dataloaders = {'train': train_loader, 'val': val_loader, 'test': test_loader}
    dataset_sizes = {'train': len(train_dataset), 'val': len(val_dataset), 'test': len(test_dataset)}

    return dataloaders, dataset_sizes, idx_to_class


# --- 4. MODEL DEFINITION (ResNet152) ---
def create_resnet152_model(pretrained=True):
    """Creates a ResNet152 model with a modified classifier."""
    print("\nInitializing ResNet152 model...")
    weights = ResNet152_Weights.IMAGENET1K_V1 if pretrained else None
    model = models.resnet152(weights=weights)

    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, config.NUM_CLASSES)
    print(f"New classifier created for {config.NUM_CLASSES} classes.")

    return model


# --- 5. TRAINING LOOP ---
def train_model(model, criterion, optimizer, scheduler, dataloaders, dataset_sizes):
    """The main training and validation loop."""
    start_time = time.time()

    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0
    epochs_no_improve = 0

    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(config.NUM_EPOCHS):
        print(f'\nEpoch {epoch+1}/{config.NUM_EPOCHS}')
        print('-' * 10)

        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()
            else:
                model.eval()

            running_loss = 0.0
            running_corrects = 0

            for inputs, labels in tqdm(dataloaders[phase], desc=f"{phase} phase"):
                inputs = inputs.to(config.DEVICE)
                labels = labels.to(config.DEVICE)

                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            if phase == 'train':
                scheduler.step()

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]

            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.cpu().numpy())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.cpu().numpy())

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())
                torch.save(model.state_dict(), os.path.join(config.OUTPUT_DIR, 'best_model.pth'))
                epochs_no_improve = 0
                print("New best validation accuracy! Saving model.")
            elif phase == 'val':
                epochs_no_improve += 1

        if epochs_no_improve >= config.PATIENCE:
            print(f'\nEarly stopping triggered after {config.PATIENCE} epochs with no improvement.')
            break

    time_elapsed = time.time() - start_time
    print(f'\nTraining complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Best val Acc: {best_acc:4f}')

    model.load_state_dict(best_model_wts)
    return model, history


# --- 6. EVALUATION ---
def evaluate_model(model, test_loader, idx_to_class):
    """Evaluates the final model on the held-out test set."""
    print("\n--- Evaluating on Test Set ---")
    model.eval()

    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc="Testing"):
            inputs = inputs.to(config.DEVICE)
            labels = labels.to(config.DEVICE)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    class_names = [idx_to_class[i] for i in range(config.NUM_CLASSES)]
    report = classification_report(all_labels, all_preds, target_names=class_names)
    print("\nClassification Report:")
    print(report)

    cm = confusion_matrix(all_labels, all_preds)
    df_cm = pd.DataFrame(cm, index=class_names, columns=class_names)

    plt.figure(figsize=(12, 8))
    sns.heatmap(df_cm, annot=True, fmt='d', cmap='Blues')
    plt.title('Confusion Matrix')
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.savefig(os.path.join(config.OUTPUT_DIR, 'confusion_matrix.png'))
    plt.show()


# --- 7. INFERENCE ON RANDOM IMAGES ---
def predict_on_random_images(model, image_paths, true_labels, idx_to_class):
    """
    Loads a trained model, makes predictions on random images,
    and displays each image with its true and predicted labels.
    """
    print("\n--- Predicting on Random Images ---")
    model.eval()

    transform = get_val_test_transforms()

    for img_path, true_label_idx in zip(image_paths, true_labels):
        try:
            image_display = Image.open(img_path).convert("RGB")
            image_np = np.array(image_display)
            transformed_image = transform(image=image_np)["image"]
            input_tensor = transformed_image.unsqueeze(0).to(config.DEVICE)

            with torch.no_grad():
                output = model(input_tensor)
                probabilities = torch.nn.functional.softmax(output[0], dim=0)
                _, pred_idx = torch.max(output, 1)

            predicted_class = idx_to_class[pred_idx.item()]
            true_class = idx_to_class[true_label_idx]
            confidence = probabilities[pred_idx.item()].item() * 100

            plt.figure(figsize=(6, 6))
            plt.imshow(image_display)
            plt.title(f"Predicted: {predicted_class} ({confidence:.2f}%)\nTrue: {true_class}")
            plt.axis('off')
            plt.show()

            print(f"Image Path: {img_path}")
            folder_name = os.path.basename(os.path.dirname(img_path))
            print(f"From Folder (True Label): {folder_name}")
            print(f"Predicted Blood Group: {predicted_class} with {confidence:.2f}% confidence.\n")

        except FileNotFoundError:
            print(f"Error: Image not found at {img_path}")
        except Exception as e:
            print(f"An error occurred while processing {img_path}: {e}")


# --- 8. MAIN EXECUTION BLOCK ---
if __name__ == '__main__':
    # --- STEP 1: PREPARE DATALOADERS ---
    dataloaders, dataset_sizes, idx_to_class = prepare_dataloaders()

    # --- STEP 2: INITIALIZE MODEL, LOSS, AND OPTIMIZER ---
    model = create_resnet152_model().to(config.DEVICE)

    # Loss function no longer needs weights due to oversampling
    criterion = nn.CrossEntropyLoss()

    optimizer = optim.Adam(model.parameters(), lr=config.LEARNING_RATE)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.1)

    # --- STEP 3: TRAIN THE MODEL ---
    trained_model, history = train_model(model, criterion, optimizer, scheduler, dataloaders, dataset_sizes)

    # --- STEP 4: EVALUATE THE FINAL MODEL ---
    evaluate_model(trained_model, dataloaders['test'], idx_to_class)

    # --- STEP 5: TEST ON RANDOM IMAGES ---
    test_dataset = dataloaders['test'].dataset
    all_test_image_paths = test_dataset.image_paths
    all_test_labels = test_dataset.labels

    num_samples = min(5, len(all_test_image_paths))
    random_indices = random.sample(range(len(all_test_image_paths)), num_samples)

    random_image_paths = [all_test_image_paths[i] for i in random_indices]
    random_true_labels = [all_test_labels[i] for i in random_indices]

    print("\nLoading best model for inference on new images...")
    inference_model = create_resnet152_model(pretrained=False).to(config.DEVICE)
    inference_model.load_state_dict(torch.load(os.path.join(config.OUTPUT_DIR, 'OVERSAMPLING_RESNET101.pth')))

    predict_on_random_images(inference_model, random_image_paths, random_true_labels, idx_to_class)

    print("\nProcess finished.")

```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms
from torchvision.models import ResNet152_Weights

import os
import numpy as np
import time
import copy
import random
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample # ADDED for oversampling
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Albumentations for advanced data augmentation
import albumentations as A
from albumentations.pytorch import ToTensorV2

# Mount Google Drive to access the dataset
try:
    from google.colab import drive
    drive.mount('/content/drive')
    IN_COLAB = True
    print("Google Drive mounted successfully.")
except ImportError:
    IN_COLAB = False
    print("Not running in Google Colab. Assuming local file paths.")

# --- CONFIGURATION (Copy from original cell) ---
class Config:
    # --- IMPORTANT: UPDATE THESE PATHS ---
    # Set the path to the root folder of your dataset in Google Drive
    DATA_DIR = '/content/drive/MyDrive/major/dataset_blood_group'

    # Set the path to save model checkpoints and results
    OUTPUT_DIR = '/content/drive/MyDrive/DATASET'

    # --- Model & Training Parameters ---
    NUM_CLASSES = 8  # A+, A-, B+, B-, AB+, AB-, O+, O-
    BATCH_SIZE = 8
    NUM_EPOCHS = 50
    LEARNING_RATE = 1e-4

    # --- Dataset Parameters ---
    CLASS_NAMES = ['A+', 'A-', 'B+', 'B-', 'AB+', 'AB-', 'O+', 'O-']
    IMAGE_SIZE = 224

    # --- Other Settings ---
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    PATIENCE = 5

# Instantiate the configuration
config = Config()

# Create output directory if it doesn't exist
os.makedirs(config.OUTPUT_DIR, exist_ok=True)
print(f"Using device: {config.DEVICE}")
print(f"Output will be saved to: {config.OUTPUT_DIR}")

# --- DATA AUGMENTATION & TRANSFORMATION (Copy from original cell) ---
def get_train_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.RandomRotate90(p=0.5),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.ElasticTransform(alpha=1, sigma=50, alpha_affine=50, p=0.5),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

def get_val_test_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

# --- DATASET & DATALOADERS (Copy from original cell) ---
class FingerprintDataset(Dataset):
    """Custom PyTorch Dataset for loading fingerprint images."""
    def __init__(self, image_paths, labels, class_to_idx, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.class_to_idx = class_to_idx

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        image = Image.open(image_path).convert("RGB")
        image = np.array(image)
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image=image)["image"]

        return image, label

def prepare_dataloaders():
    """Scans the data directory, oversamples the entire dataset, splits it, and creates DataLoaders."""
    print("Preparing dataloaders...")

    if not os.path.isdir(config.DATA_DIR):
        raise FileNotFoundError(f"Data directory not found at: {config.DATA_DIR}. Please update the path in the Config class.")

    # Create a DataFrame of all images and labels
    filepaths = []
    labels = []
    for class_name in config.CLASS_NAMES:
        class_dir = os.path.join(config.DATA_DIR, class_name)
        if not os.path.isdir(class_dir):
            continue
        for img_name in os.listdir(class_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
                filepaths.append(os.path.join(class_dir, img_name))
                labels.append(class_name)

    data = pd.DataFrame({'Filepath': filepaths, 'Label': labels})

    # --- VISUALIZE ORIGINAL DISTRIBUTION ---
    print("\n--- Original Full Dataset Distribution ---")
    print(data['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Original Full Dataset')
    plt.ylabel('Number of Images')
    plt.show()

    # --- OVERSAMPLING LOGIC (APPLIED TO ENTIRE DATASET) ---
    # WARNING: This approach can lead to data leakage and overly optimistic results.
    # It is generally recommended to split the data first, then oversample only the training set.
    print("\n--- Applying Oversampling to Entire Dataset ---")
    max_count = data['Label'].value_counts().max()
    print(f"The majority class has {max_count} samples. Oversampling minority classes to match this count.")

    class_dfs = [data[data['Label'] == label] for label in config.CLASS_NAMES]

    oversampled_dfs = [
        resample(df, replace=True, n_samples=max_count, random_state=42) if len(df) < max_count else df
        for df in class_dfs
    ]

    data_oversampled = pd.concat(oversampled_dfs)
    data_oversampled = data_oversampled.sample(frac=1, random_state=42).reset_index(drop=True)

    # --- VISUALIZE BALANCED DISTRIBUTION ---
    print("\n--- Full Dataset Distribution After Oversampling ---")
    print(data_oversampled['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data_oversampled, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Full Dataset After Oversampling')
    plt.ylabel('Number of Images')
    plt.show()

    # Split the now-oversampled data into training, validation, and test sets
    train_df, test_df = train_test_split(data_oversampled, test_size=0.30, random_state=42, stratify=data_oversampled['Label'])
    val_df, test_df = train_test_split(test_df, test_size=0.50, random_state=42, stratify=test_df['Label'])

    # Get file paths and labels from the dataframes
    class_to_idx = {name: i for i, name in enumerate(config.CLASS_NAMES)}
    idx_to_class = {i: name for i, name in enumerate(config.CLASS_NAMES)}

    train_paths = train_df['Filepath'].tolist()
    train_labels = [class_to_idx[label] for label in train_df['Label']]

    val_paths = val_df['Filepath'].tolist()
    val_labels = [class_to_idx[label] for label in val_df['Label']]

    test_paths = test_df['Filepath'].tolist()
    test_labels = [class_to_idx[label] for label in test_df['Label']]

    print(f"\nTraining set size: {len(train_paths)}")
    print(f"Validation set size: {len(val_paths)}")
    print(f"Test set size: {len(test_paths)}")

    # Create datasets
    train_dataset = FingerprintDataset(train_paths, train_labels, class_to_idx, transform=get_train_transforms())
    # CORRECTED: Pass class_to_idx for validation and test datasets
    val_dataset = FingerprintDataset(val_paths, val_labels, class_to_idx, transform=get_val_test_transforms())
    test_dataset = FingerprintDataset(test_paths, test_labels, class_to_idx, transform=get_val_test_transforms())


    # Create dataloaders (no sampler needed now)
    train_loader = DataLoader(train_dataset, batch_size=config.BATCH_SIZE, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)


    dataloaders = {'train': train_loader, 'val': val_loader, 'test': test_loader}
    dataset_sizes = {'train': len(train_dataset), 'val': len(val_dataset), 'test': len(test_dataset)}

    return dataloaders, dataset_sizes, idx_to_class

# --- MODEL DEFINITION (ResNet152) (Copy from original cell) ---
def create_resnet152_model(pretrained=True):
    """Creates a ResNet152 model with a modified classifier."""
    print("\nInitializing ResNet152 model...")
    weights = ResNet152_Weights.IMAGENET1K_V1 if pretrained else None
    model = models.resnet152(weights=weights)

    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, config.NUM_CLASSES)
    print(f"New classifier created for {config.NUM_CLASSES} classes.")

    return model


# --- 7. INFERENCE ON RANDOM IMAGES (MODIFIED) ---
def predict_on_random_images_modified(model, image_paths, true_labels, idx_to_class):
    """
    Loads a trained model, makes predictions on random images,
    and displays each image with its true and predicted labels.
    Modified to load 'best_model.pth'.
    """
    print("\n--- Predicting on Random Images ---")
    model.eval()

    transform = get_val_test_transforms()

    for img_path, true_label_idx in zip(image_paths, true_labels):
        try:
            image_display = Image.open(img_path).convert("RGB")
            image_np = np.array(image_display)
            transformed_image = transform(image=image_np)["image"]
            input_tensor = transformed_image.unsqueeze(0).to(config.DEVICE)

            with torch.no_grad():
                output = model(input_tensor)
                probabilities = torch.nn.functional.softmax(output[0], dim=0)
                _, pred_idx = torch.max(output, 1)

            predicted_class = idx_to_class[pred_idx.item()]
            true_class = idx_to_class[true_label_idx]
            confidence = probabilities[pred_idx.item()].item() * 100

            plt.figure(figsize=(6, 6))
            plt.imshow(image_display)
            plt.title(f"Predicted: {predicted_class} ({confidence:.2f}%)\nTrue: {true_class}")
            plt.axis('off')
            plt.show()

            print(f"Image Path: {img_path}")
            folder_name = os.path.basename(os.path.dirname(img_path))
            print(f"From Folder (True Label): {folder_name}")
            print(f"Predicted Blood Group: {predicted_class} with {confidence:.2f}% confidence.\n")

        except FileNotFoundError:
            print(f"Error: Image not found at {img_path}")
        except Exception as e:
            print(f"An error occurred while processing {img_path}: {e}")

# --- 8. MAIN EXECUTION BLOCK (MODIFIED FOR INFERENCE ONLY) ---
if __name__ == '__main__':
    # --- STEP 1: PREPARE DATALOADERS ---
    # This step is still needed to get dataset information and idx_to_class
    dataloaders, dataset_sizes, idx_to_class = prepare_dataloaders()

    # --- STEP 5: TEST ON RANDOM IMAGES ---
    test_dataset = dataloaders['test'].dataset
    all_test_image_paths = test_dataset.image_paths
    all_test_labels = test_dataset.labels

    num_samples = min(5, len(all_test_image_paths))
    random_indices = random.sample(range(len(all_test_image_paths)), num_samples)

    random_image_paths = [all_test_image_paths[i] for i in random_indices]
    random_true_labels = [all_test_labels[i] for i in random_indices]

    print("\nLoading best model for inference on new images...")
    inference_model = create_resnet152_model(pretrained=False).to(config.DEVICE)
    # MODIFIED: Load the best_model.pth file with map_location
    inference_model.load_state_dict(torch.load(os.path.join(config.OUTPUT_DIR, 'best_model.pth'), map_location=torch.device('cpu')))

    # Use the modified inference function
    predict_on_random_images_modified(inference_model, random_image_paths, random_true_labels, idx_to_class)

    print("\nProcess finished.")
```

```python
# ==============================================================================
# Blood Group Prediction from Fingerprints: Complete Training Notebook
# ==============================================================================
# This script is designed for Google Colab. It uses scikit-learn's resample
# function to oversample the ENTIRE dataset BEFORE splitting.
#
# It implements:
# - ResNet152 with transfer learning
# - Advanced, domain-specific data augmentation (Elastic Distortions)
# - Handling of severe class imbalance (Oversampling with sklearn.utils.resample)
# - A complete training, validation, and evaluation pipeline.
# - An inference function to test on new, random images.
# ==============================================================================

# --- 0. SETUP AND IMPORTS ---
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms
from torchvision.models import ResNet152_Weights

import os
import numpy as np
import time
import copy
import random
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample # ADDED for oversampling
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Albumentations for advanced data augmentation
import albumentations as A
from albumentations.pytorch import ToTensorV2

# Mount Google Drive to access the dataset
try:
    from google.colab import drive
    drive.mount('/content/drive')
    IN_COLAB = True
    print("Google Drive mounted successfully.")
except ImportError:
    IN_COLAB = False
    print("Not running in Google Colab. Assuming local file paths.")


# --- 1. CONFIGURATION ---
class Config:
    # --- IMPORTANT: UPDATE THESE PATHS ---
    # Set the path to the root folder of your dataset in Google Drive
    DATA_DIR = '/content/drive/MyDrive/major/dataset_blood_group'

    # Set the path to save model checkpoints and results
    OUTPUT_DIR = '/content/drive/MyDrive/DATASET'

    # --- Model & Training Parameters ---
    NUM_CLASSES = 8  # A+, A-, B+, B-, AB+, AB-, O+, O-
    BATCH_SIZE = 8
    NUM_EPOCHS = 50
    LEARNING_RATE = 1e-4

    # --- Dataset Parameters ---
    CLASS_NAMES = ['A+', 'A-', 'B+', 'B-', 'AB+', 'AB-', 'O+', 'O-']
    IMAGE_SIZE = 224

    # --- Other Settings ---
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    PATIENCE = 5

# Instantiate the configuration
config = Config()

# Create output directory if it doesn't exist
os.makedirs(config.OUTPUT_DIR, exist_ok=True)
print(f"Using device: {config.DEVICE}")
print(f"Output will be saved to: {config.OUTPUT_DIR}")


# --- 2. DATA AUGMENTATION & TRANSFORMATION ---
def get_train_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.RandomRotate90(p=0.5),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.ElasticTransform(alpha=1, sigma=50, alpha_affine=50, p=0.5),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

def get_val_test_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])


# --- 3. DATASET & DATALOADERS (UPDATED FOR OVERSAMPLING BEFORE SPLIT) ---
class FingerprintDataset(Dataset):
    """Custom PyTorch Dataset for loading fingerprint images."""
    def __init__(self, image_paths, labels, class_to_idx, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.class_to_idx = class_to_idx

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        image = Image.open(image_path).convert("RGB")
        image = np.array(image)
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image=image)["image"]

        return image, label

def prepare_dataloaders():
    """Scans the data directory, oversamples the entire dataset, splits it, and creates DataLoaders."""
    print("Preparing dataloaders...")

    if not os.path.isdir(config.DATA_DIR):
        raise FileNotFoundError(f"Data directory not found at: {config.DATA_DIR}. Please update the path in the Config class.")

    # Create a DataFrame of all images and labels
    filepaths = []
    labels = []
    for class_name in config.CLASS_NAMES:
        class_dir = os.path.join(config.DATA_DIR, class_name)
        if not os.path.isdir(class_dir):
            continue
        for img_name in os.listdir(class_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
                filepaths.append(os.path.join(class_dir, img_name))
                labels.append(class_name)

    data = pd.DataFrame({'Filepath': filepaths, 'Label': labels})

    # --- VISUALIZE ORIGINAL DISTRIBUTION ---
    print("\n--- Original Full Dataset Distribution ---")
    print(data['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Original Full Dataset')
    plt.ylabel('Number of Images')
    plt.show()

    # --- OVERSAMPLING LOGIC (APPLIED TO ENTIRE DATASET) ---
    # WARNING: This approach can lead to data leakage and overly optimistic results.
    # It is generally recommended to split the data first, then oversample only the training set.
    print("\n--- Applying Oversampling to Entire Dataset ---")
    max_count = data['Label'].value_counts().max()
    print(f"The majority class has {max_count} samples. Oversampling minority classes to match this count.")

    class_dfs = [data[data['Label'] == label] for label in config.CLASS_NAMES]

    oversampled_dfs = [
        resample(df, replace=True, n_samples=max_count, random_state=42) if len(df) < max_count else df
        for df in class_dfs
    ]

    data_oversampled = pd.concat(oversampled_dfs)
    data_oversampled = data_oversampled.sample(frac=1, random_state=42).reset_index(drop=True)

    # --- VISUALIZE BALANCED DISTRIBUTION ---
    print("\n--- Full Dataset Distribution After Oversampling ---")
    print(data_oversampled['Label'].value_counts())
    plt.figure(figsize=(10, 6))
    sns.countplot(x='Label', data=data_oversampled, order=config.CLASS_NAMES)
    plt.title('Class Distribution in Full Dataset After Oversampling')
    plt.ylabel('Number of Images')
    plt.show()

    # Split the now-oversampled data into training, validation, and test sets
    train_df, test_df = train_test_split(data_oversampled, test_size=0.30, random_state=42, stratify=data_oversampled['Label'])
    val_df, test_df = train_test_split(test_df, test_size=0.50, random_state=42, stratify=test_df['Label'])

    # Get file paths and labels from the dataframes
    class_to_idx = {name: i for i, name in enumerate(config.CLASS_NAMES)}
    idx_to_class = {i: name for i, name in enumerate(config.CLASS_NAMES)}

    train_paths = train_df['Filepath'].tolist()
    train_labels = [class_to_idx[label] for label in train_df['Label']]

    val_paths = val_df['Filepath'].tolist()
    val_labels = [class_to_idx[label] for label in val_df['Label']]

    test_paths = test_df['Filepath'].tolist()
    test_labels = [class_to_idx[label] for label in test_df['Label']]

    print(f"\nTraining set size: {len(train_paths)}")
    print(f"Validation set size: {len(val_paths)}")
    print(f"Test set size: {len(test_paths)}")

    # Create datasets
    train_dataset = FingerprintDataset(train_paths, train_labels, class_to_idx, transform=get_train_transforms())
    # CORRECTED: Pass val_labels and test_labels as the second argument
    val_dataset = FingerprintDataset(val_paths, val_labels, class_to_idx, transform=get_val_test_transforms())
    test_dataset = FingerprintDataset(test_paths, test_labels, class_to_idx, transform=get_val_test_transforms())

    # Create dataloaders (no sampler needed now)
    train_loader = DataLoader(train_dataset, batch_size=config.BATCH_SIZE, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)

    dataloaders = {'train': train_loader, 'val': val_loader, 'test': test_loader}
    dataset_sizes = {'train': len(train_dataset), 'val': len(val_dataset), 'test': len(test_dataset)}

    return dataloaders, dataset_sizes, idx_to_class


# --- 4. MODEL DEFINITION (ResNet152) ---
def create_resnet152_model(pretrained=True):
    """Creates a ResNet152 model with a modified classifier."""
    print("\nInitializing ResNet152 model...")
    weights = ResNet152_Weights.IMAGENET1K_V1 if pretrained else None
    model = models.resnet152(weights=weights)

    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, config.NUM_CLASSES)
    print(f"New classifier created for {config.NUM_CLASSES} classes.")

    return model


# --- 5. TRAINING LOOP (MODIFIED FOR RESUMING) ---
def train_model(model, criterion, optimizer, scheduler, dataloaders, dataset_sizes, start_epoch=0):
    """The main training and validation loop."""
    start_time = time.time()

    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0
    epochs_no_improve = 0

    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(start_epoch, config.NUM_EPOCHS): # Start from start_epoch
        print(f'\nEpoch {epoch+1}/{config.NUM_EPOCHS}')
        print('-' * 10)

        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()
            else:
                model.eval()

            running_loss = 0.0
            running_corrects = 0

            for inputs, labels in tqdm(dataloaders[phase], desc=f"{phase} phase"):
                inputs = inputs.to(config.DEVICE)
                labels = labels.to(config.DEVICE)

                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            if phase == 'train':
                scheduler.step()

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]

            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.cpu().numpy())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.cpu().numpy())

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())
                torch.save(model.state_dict(), os.path.join(config.OUTPUT_DIR, 'best_model.pth'))
                epochs_no_improve = 0
                print("New best validation accuracy! Saving model.")
            elif phase == 'val':
                epochs_no_improve += 1

        if epochs_no_improve >= config.PATIENCE:
            print(f'\nEarly stopping triggered after {config.PATIENCE} epochs with no improvement.')
            break

    time_elapsed = time.time() - start_time
    print(f'\nTraining complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Best val Acc: {best_acc:4f}')

    model.load_state_dict(best_model_wts)
    return model, history


# --- 6. EVALUATION ---
def evaluate_model(model, test_loader, idx_to_class):
    """Evaluates the final model on the held-out test set."""
    print("\n--- Evaluating on Test Set ---")
    model.eval()

    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc="Testing"):
            inputs = inputs.to(config.DEVICE)
            labels = labels.to(config.DEVICE)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    class_names = [idx_to_class[i] for i in range(config.NUM_CLASSES)]
    report = classification_report(all_labels, all_preds, target_names=class_names)
    print("\nClassification Report:")
    print(report)

    cm = confusion_matrix(all_labels, all_preds)
    df_cm = pd.DataFrame(cm, index=class_names, columns=class_names)

    plt.figure(figsize=(12, 8))
    sns.heatmap(df_cm, annot=True, fmt='d', cmap='Blues')
    plt.title('Confusion Matrix')
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.savefig(os.path.join(config.OUTPUT_DIR, 'confusion_matrix.png'))
    plt.show()


# --- 7. INFERENCE ON RANDOM IMAGES ---
def predict_on_random_images(model, image_paths, true_labels, idx_to_class):
    """
    Loads a trained model, makes predictions on random images,
    and displays each image with its true and predicted labels.
    """
    print("\n--- Predicting on Random Images ---")
    model.eval()

    transform = get_val_test_transforms()

    for img_path, true_label_idx in zip(image_paths, true_labels):
        try:
            image_display = Image.open(img_path).convert("RGB")
            image_np = np.array(image_display)
            transformed_image = transform(image=image_np)["image"]
            input_tensor = transformed_image.unsqueeze(0).to(config.DEVICE)

            with torch.no_grad():
                output = model(input_tensor)
                probabilities = torch.nn.functional.softmax(output[0], dim=0)
                _, pred_idx = torch.max(output, 1)

            predicted_class = idx_to_class[pred_idx.item()]
            true_class = idx_to_class[true_label_idx]
            confidence = probabilities[pred_idx.item()].item() * 100

            plt.figure(figsize=(6, 6))
            plt.imshow(image_display)
            plt.title(f"Predicted: {predicted_class} ({confidence:.2f}%)\nTrue: {true_class}")
            plt.axis('off')
            plt.show()

            print(f"Image Path: {img_path}")
            folder_name = os.path.basename(os.path.dirname(img_path))
            print(f"From Folder (True Label): {folder_name}")
            print(f"Predicted Blood Group: {predicted_class} with {confidence:.2f}% confidence.\n")

        except FileNotFoundError:
            print(f"Error: Image not found at {img_path}")
        except Exception as e:
            print(f"An error occurred while processing {img_path}: {e}")


# --- 8. MAIN EXECUTION BLOCK (MODIFIED FOR RESUMING) ---
if __name__ == '__main__':
    # --- STEP 1: PREPARE DATALOADERS ---
    dataloaders, dataset_sizes, idx_to_class = prepare_dataloaders()

    # --- STEP 2: INITIALIZE MODEL, LOSS, AND OPTIMIZER ---
    model = create_resnet152_model().to(config.DEVICE)

    # Load the best model state if it exists to resume training
    model_save_path = os.path.join(config.OUTPUT_DIR, 'best_model.pth')
    start_epoch = 0
    if os.path.exists(model_save_path):
        print(f"Loading model from {model_save_path} to resume training...")
        model.load_state_dict(torch.load(model_save_path, map_location=config.DEVICE))
        # Note: To precisely resume from the last checkpoint, you would ideally save
        # the optimizer state, scheduler state, and epoch number as well.
        # This modification only loads the model weights.
        # We'll manually set start_epoch to continue the loop, assuming the best model
        # was saved around epoch 19 as you mentioned. You might need to adjust this.
        start_epoch = 19 # Manually set the starting epoch

    # Loss function no longer needs weights due to oversampling
    criterion = nn.CrossEntropyLoss()

    optimizer = optim.Adam(model.parameters(), lr=config.LEARNING_RATE)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.1)

    # --- STEP 3: TRAIN THE MODEL ---
    # Pass the starting epoch to the training loop
    trained_model, history = train_model(model, criterion, optimizer, scheduler, dataloaders, dataset_sizes, start_epoch=start_epoch)

    # --- STEP 4: EVALUATE THE FINAL MODEL ---
    evaluate_model(trained_model, dataloaders['test'], idx_to_class)

    # --- STEP 5: TEST ON RANDOM IMAGES ---
    test_dataset = dataloaders['test'].dataset
    all_test_image_paths = test_dataset.image_paths
    all_test_labels = test_dataset.labels

    num_samples = min(5, len(all_test_image_paths))
    random_indices = random.sample(range(len(all_test_image_paths)), num_samples)

    random_image_paths = [all_test_image_paths[i] for i in random_indices]
    random_true_labels = [all_test_labels[i] for i in random_indices]

    print("\nLoading best model for inference on new images...")
    # Create a new model instance for inference and load the best weights
    inference_model = create_resnet152_model(pretrained=False).to(config.DEVICE)
    inference_model.load_state_dict(torch.load(os.path.join(config.OUTPUT_DIR, 'best_model.pth'), map_location=config.DEVICE))

    predict_on_random_images(inference_model, random_image_paths, random_true_labels, idx_to_class)

    print("\nProcess finished.")
```

```python
# ==============================================================================
# Blood Group Prediction from Fingerprints: Evaluation and Inference
# ==============================================================================
# This script is designed for Google Colab. It loads a pre-trained model,
# evaluates it on the test set, and performs inference on random images.
#
# It assumes:
# - The dataset is mounted in Google Drive at the specified path.
# - A trained model checkpoint ('best_model.pth') exists in the output directory.
# ==============================================================================

# --- 0. SETUP AND IMPORTS ---
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms
from torchvision.models import ResNet152_Weights

import os
import numpy as np
import time
import copy
import random
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample # ADDED for oversampling
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Albumentations for advanced data augmentation
import albumentations as A
from albumentations.pytorch import ToTensorV2

# Mount Google Drive to access the dataset
try:
    from google.colab import drive
    drive.mount('/content/drive')
    IN_COLAB = True
    print("Google Drive mounted successfully.")
except ImportError:
    IN_COLAB = False
    print("Not running in Google Colab. Assuming local file paths.")

# --- CONFIGURATION (Copy from original cell) ---
class Config:
    # --- IMPORTANT: UPDATE THESE PATHS ---
    # Set the path to the root folder of your dataset in Google Drive
    DATA_DIR = '/content/drive/MyDrive/major/dataset_blood_group'

    # Set the path to save model checkpoints and results
    OUTPUT_DIR = '/content/drive/MyDrive/DATASET'

    # --- Model & Training Parameters ---
    NUM_CLASSES = 8  # A+, A-, B+, B-, AB+, AB-, O+, O-
    BATCH_SIZE = 8
    NUM_EPOCHS = 50 # This will not be used in this script, but kept for consistency
    LEARNING_RATE = 1e-4 # This will not be used in this script

    # --- Dataset Parameters ---
    CLASS_NAMES = ['A+', 'A-', 'B+', 'B-', 'AB+', 'AB-', 'O+', 'O-']
    IMAGE_SIZE = 224

    # --- Other Settings ---
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    PATIENCE = 5 # This will not be used in this script

# Instantiate the configuration
config = Config()

# Create output directory if it doesn't exist
os.makedirs(config.OUTPUT_DIR, exist_ok=True)
print(f"Using device: {config.DEVICE}")
print(f"Output will be saved to: {config.OUTPUT_DIR}")

# --- DATA AUGMENTATION & TRANSFORMATION (Copy from original cell) ---
def get_train_transforms():
    # This function is not strictly needed for inference/evaluation, but kept for completeness
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.RandomRotate90(p=0.5),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.ElasticTransform(alpha=1, sigma=50, alpha_affine=50, p=0.5),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

def get_val_test_transforms():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
        ToTensorV2(),
    ])

# --- 3. DATASET & DATALOADERS (Copy from original cell - simplified for test data only) ---
class FingerprintDataset(Dataset):
    """Custom PyTorch Dataset for loading fingerprint images."""
    def __init__(self, image_paths, labels, class_to_idx, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.class_to_idx = class_to_idx

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        image = Image.open(image_path).convert("RGB")
        image = np.array(image)
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image=image)["image"]

        return image, label

def prepare_test_dataloader():
    """Scans the data directory, prepares test data, and creates a DataLoader."""
    print("Preparing test dataloader...")

    if not os.path.isdir(config.DATA_DIR):
        raise FileNotFoundError(f"Data directory not found at: {config.DATA_DIR}. Please update the path in the Config class.")

    # Create a DataFrame of all images and labels
    filepaths = []
    labels = []
    for class_name in config.CLASS_NAMES:
        class_dir = os.path.join(config.DATA_DIR, class_name)
        if not os.path.isdir(class_dir):
            continue
        for img_name in os.listdir(class_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif')):
                filepaths.append(os.path.join(class_dir, img_name))
                labels.append(class_name)

    data = pd.DataFrame({'Filepath': filepaths, 'Label': labels})

    # --- OVERSAMPLING LOGIC (APPLIED TO ENTIRE DATASET - NEEDED FOR CONSISTENT SPLIT) ---
    # Although we only use the test set here, the oversampling needs to be
    # applied consistently before splitting to reproduce the original test set.
    print("\n--- Applying Oversampling to Entire Dataset (for consistent split) ---")
    max_count = data['Label'].value_counts().max()
    print(f"The majority class has {max_count} samples. Oversampling minority classes to match this count.")

    class_dfs = [data[data['Label'] == label] for label in config.CLASS_NAMES]

    oversampled_dfs = [
        resample(df, replace=True, n_samples=max_count, random_state=42) if len(df) < max_count else df
        for df in class_dfs
    ]

    data_oversampled = pd.concat(oversampled_dfs)
    data_oversampled = data_oversampled.sample(frac=1, random_state=42).reset_index(drop=True)

    # Split the now-oversampled data into training, validation, and test sets
    # We only need the test_df for this script
    train_df, test_df = train_test_split(data_oversampled, test_size=0.30, random_state=42, stratify=data_oversampled['Label'])
    val_df, test_df = train_test_split(test_df, test_size=0.50, random_state=42, stratify=test_df['Label'])

    # Get file paths and labels from the test dataframe
    class_to_idx = {name: i for i, name in enumerate(config.CLASS_NAMES)}
    idx_to_class = {i: name for i, name in enumerate(config.CLASS_NAMES)}

    test_paths = test_df['Filepath'].tolist()
    test_labels = [class_to_idx[label] for label in test_df['Label']]

    print(f"\nTest set size: {len(test_paths)}")

    # Create test dataset and dataloader
    test_dataset = FingerprintDataset(test_paths, test_labels, class_to_idx, transform=get_val_test_transforms())
    test_loader = DataLoader(test_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=2)

    return test_loader, len(test_dataset), idx_to_class


# --- 4. MODEL DEFINITION (ResNet152) (Copy from original cell) ---
def create_resnet152_model(pretrained=True):
    """Creates a ResNet152 model with a modified classifier."""
    print("\nInitializing ResNet152 model...")
    weights = ResNet152_Weights.IMAGENET1K_V1 if pretrained else None
    model = models.resnet152(weights=weights)

    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, config.NUM_CLASSES)
    print(f"New classifier created for {config.NUM_CLASSES} classes.")

    return model


# --- 6. EVALUATION (Copy from original cell) ---
def evaluate_model(model, test_loader, idx_to_class):
    """Evaluates the final model on the held-out test set."""
    print("\n--- Evaluating on Test Set ---")
    model.eval()

    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc="Testing"):
            inputs = inputs.to(config.DEVICE)
            labels = labels.to(config.DEVICE)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    class_names = [idx_to_class[i] for i in range(config.NUM_CLASSES)]
    report = classification_report(all_labels, all_preds, target_names=class_names)
    print("\nClassification Report:")
    print(report)

    cm = confusion_matrix(all_labels, all_preds)
    df_cm = pd.DataFrame(cm, index=class_names, columns=class_names)

    plt.figure(figsize=(12, 8))
    sns.heatmap(df_cm, annot=True, fmt='d', cmap='Blues')
    plt.title('Confusion Matrix')
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.savefig(os.path.join(config.OUTPUT_DIR, 'confusion_matrix.png'))
    plt.show()


# --- 7. INFERENCE ON RANDOM IMAGES (Copy from original cell) ---
def predict_on_random_images(model, image_paths, true_labels, idx_to_class):
    """
    Loads a trained model, makes predictions on random images from the test set,
    and displays each image with its true and predicted labels.
    """
    print("\n--- Predicting on Random Images ---")
    model.eval()

    transform = get_val_test_transforms()

    # Select a few random images from the provided paths
    num_samples = min(5, len(image_paths))
    random_indices = random.sample(range(len(image_paths)), num_samples)
    random_image_paths = [image_paths[i] for i in random_indices]
    random_true_labels = [true_labels[i] for i in random_indices]


    for img_path, true_label_idx in zip(random_image_paths, random_true_labels):
        try:
            image_display = Image.open(img_path).convert("RGB")
            image_np = np.array(image_display)
            transformed_image = transform(image=image_np)["image"]
            input_tensor = transformed_image.unsqueeze(0).to(config.DEVICE)

            with torch.no_grad():
                output = model(input_tensor)
                probabilities = torch.nn.functional.softmax(output[0], dim=0)
                _, pred_idx = torch.max(output, 1)

            predicted_class = idx_to_class[pred_idx.item()]
            true_class = idx_to_class[true_label_idx]
            confidence = probabilities[pred_idx.item()].item() * 100

            plt.figure(figsize=(6, 6))
            plt.imshow(image_display)
            plt.title(f"Predicted: {predicted_class} ({confidence:.2f}%)\nTrue: {true_class}")
            plt.axis('off')
            plt.show()

            print(f"Image Path: {img_path}")
            folder_name = os.path.basename(os.path.dirname(img_path))
            print(f"From Folder (True Label): {folder_name}")
            print(f"Predicted Blood Group: {predicted_class} with {confidence:.2f}% confidence.\n")

        except FileNotFoundError:
            print(f"Error: Image not found at {img_path}")
        except Exception as e:
            print(f"An error occurred while processing {img_path}: {e}")


# --- 8. MAIN EXECUTION BLOCK (Evaluation and Inference Only) ---
if __name__ == '__main__':
    # --- STEP 1: PREPARE TEST DATALOADER ---
    test_loader, test_dataset_size, idx_to_class = prepare_test_dataloader()

    # --- STEP 2: INITIALIZE MODEL AND LOAD WEIGHTS ---
    model = create_resnet152_model(pretrained=False).to(config.DEVICE) # Initialize model without pretrained weights

    # Load the best model state
    model_save_path = os.path.join(config.OUTPUT_DIR, 'best_model.pth')
    if os.path.exists(model_save_path):
        print(f"Loading model weights from {model_save_path}...")
        model.load_state_dict(torch.load(model_save_path, map_location=config.DEVICE))
    else:
        raise FileNotFoundError(f"Model checkpoint not found at: {model_save_path}. Please train the model first.")


    # --- STEP 3: EVALUATE THE LOADED MODEL ---
    evaluate_model(model, test_loader, idx_to_class)

    # --- STEP 4: TEST ON RANDOM IMAGES ---
    # Get all image paths and labels from the test dataset
    all_test_image_paths = test_loader.dataset.image_paths
    all_test_labels = test_loader.dataset.labels

    predict_on_random_images(model, all_test_image_paths, all_test_labels, idx_to_class)

    print("\nProcess finished.")
```
